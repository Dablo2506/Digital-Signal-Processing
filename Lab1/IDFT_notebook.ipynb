# Discrete-time signal synthesis using the IDFT

Input spectrum: X = [6, 2, 4, 3, 4, 4, 0, 0, 0, 0]^T.

Convention: W[n,k] = exp(-j 2 pi n k / N), K = W^H / N, x = K X.
The exercise does not define the letters W and K; these names are explicitly chosen here.

N = 10 is the supplied case. N = 16 and N = 20 are illustrative additional cases using trailing zero padding of the spectrum. No real-valued constraint or conjugate symmetry is imposed.


```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([6, 2, 4, 3, 4, 4, 0, 0, 0, 0], dtype=float)
N_values = [10, 16, 20]
results = {}
fig, axes = plt.subplots(3, 1, figsize=(10, 9), constrained_layout=True)
for ax, N in zip(axes, N_values):
    XN = np.pad(X, (0, N - len(X)))
    n = np.arange(N)
    W = np.exp(-2j * np.pi * np.outer(n, n) / N)
    K = W.conj().T / N
    x = K @ XN
    assert np.allclose(K @ W, np.eye(N))
    assert np.allclose(x, np.fft.ifft(XN))
    results[N] = x
    np.set_printoptions(precision=4, linewidth=180)
    print(f"N = {N}\nW =\n{W}\nK =\n{K}\nx =\n{x}\n")
    ax.stem(n, x.real, linefmt='C0-', markerfmt='C0o',
            basefmt='k-', label='Real part')
    ax.stem(n, x.imag, linefmt='C1--', markerfmt='C1x',
            basefmt=' ', label='Imaginary part')
    ax.set(title=f'N = {N}', xlabel='Sample n', ylabel='Amplitude')
    ax.grid(alpha=0.25)
    ax.legend()
plt.show()
```

## Interpretation
The output is complex because the input spectrum is not conjugate symmetric. At n = 0 the values are 23/N: 2.3, 1.4375 and 1.15 for N = 10, 16 and 20. Keeping the coefficients fixed while increasing N changes their normalized frequencies and the 1/N scale; this is not amplitude-preserving interpolation of an unchanged waveform.
